```python
class Parent:
    def show(self):
        print("Parent class method")

class Child(Parent):
    def display(self):
        print("Child class method")

obj = Child()
obj.show()  
obj.display()  
```

```python
class Parent1:
    def method1(self):
        print("Parent1 method")

class Parent2:
    def method2(self):
        print("Parent2 method")

class Child(Parent1, Parent2):
    def method3(self):
        print("Child class method")

obj = Child()
obj.method1()   
obj.method2()   
obj.method3()   
```

```python
class Grandparent:
    def method1(self):
        print("Grandparent method")

class Parent(Grandparent):
    def method2(self):
        print("Parent method")

class Child(Parent):
    def method3(self):
        print("Child class method")

obj = Child()
obj.method1()  
obj.method2()  
obj.method3()  
```

```python
class Parent:
    def method1(self):
        print("Parent method")

class Child1(Parent):
    def method2(self):
        print("Child1 method")

class Child2(Parent):
    def method3(self):
        print("Child2 method")

obj1 = Child1()
obj2 = Child2()

obj1.method1()  
obj1.method2()  

obj2.method1()  
obj2.method3()  
```

```python
class Parent1:
    def method1(self):
        print("Parent1 method")

class Parent2:
    def method2(self):
        print("Parent2 method")

class Child1(Parent1, Parent2):  
    def method3(self):
        print("Child1 method")

class Child2(Child1):  
    def method4(self):
        print("Child2 method")

obj = Child2()
obj.method1()  
obj.method2()  
obj.method3() 
obj.method4()  
```
